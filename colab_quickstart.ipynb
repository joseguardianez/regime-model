## 6. Try your own buckets (optional)
Change the tickers below and run the cell. It uses the same functions as the script.

```python
import regime_model as m

m.BUCKETS = {
    "Disinflation": ["TSLA", "PANW", "META", "AVGO", "AMZN"],
    "Reflation":    ["XOM", "PWR", "PG", "CVX", "JNJ"],
    "Stagflation":  ["YTRA", "INTG", "SPWH", "JILL", "DOGZ"],
    "Deflation":    ["HSY", "ELA", "SMMT", "CCLD", "LNTH"],
}
m.OUTSIDE_ASSETS = ["QQQ", "AMD"]

prices, vix, outside = m.load_yahoo()
votes, raw, score, winner = m.build_regimes(prices, vix)
print("Current regime:", m.REGIMES[int(winner.dropna().iloc[-1]) - 1])
score.plot(figsize=(14, 5), title="Regime scores");
```
